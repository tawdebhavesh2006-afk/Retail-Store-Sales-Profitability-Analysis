```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv(r"C:\Users\Dell\OneDrive\Documents\DATA ANALYSIS PROJECT\Retail Store Sales & Profitability Analysis\EXCELL\retail_store_sales_profitability_raw.csv")
df
```

```python
month_order = [
    "January", "February", "March", "April",
    "May", "June", "July", "August",
    "September", "October", "November", "December"
]

df["MONTHs"] = pd.Categorical(
    df["Month"],
    categories=month_order,
    ordered=True
)

monthly_changes = df.groupby("Month", observed=True)["Sales"].sum()

monthly_changes.plot(kind="line", marker="o", figsize=(12, 6))

plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.xticks(rotation=45)
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
analysis=df.groupby('Store_Name')['Sales'].sum()
analysis.plot(kind="bar",figsize=(12, 6))

plt.title("SALES BY STORE")
plt.xlabel("STORE")
plt.ylabel("SALES")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
analysis=df.groupby('Store_Name')['Profit'].sum()
analysis.plot(kind="bar",figsize=(12, 6))

plt.title("PROFIT BY STORE")
plt.xlabel("STORE")
plt.ylabel("PROFTI")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
analysis=df.groupby('Employee_Name')['Sales'].sum().sort_values(ascending=False).head()
analysis.plot(kind="barh",figsize=(12, 6))

plt.title("EMPLOYEE GENERATE MOST SALES")
plt.xlabel("SALES")
plt.ylabel("EMPLOYEE")
plt.xticks(rotation=360)
plt.tight_layout()
plt.show()
```

```python
analysis=df.groupby('Category')['Profit'].sum().sort_values(ascending=False)
analysis.plot(kind="bar",figsize=(12, 6))

plt.title("MOST PROFITABLE CATEGORY")
plt.xlabel("CATEGORY")
plt.ylabel("PROFTI")
plt.xticks(rotation=360)
plt.tight_layout()
plt.show()
```

```python
sns.histplot(data=df,x='Sales',bins=30)
plt.title("Sales Distribution")
plt.xlabel("Sales")
plt.ylabel("Frequency")
plt.show()
```

```python
sns.regplot(
    data=df,
    x="Discount",
    y="Profit"
)

plt.title("Discount vs Profit")
plt.xlabel("Discount")
plt.ylabel("Profit")
plt.show()
```
